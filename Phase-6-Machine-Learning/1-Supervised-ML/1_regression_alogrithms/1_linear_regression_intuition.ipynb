# Linear Regression

supervised machine learning algorithm used to model the relationship between a dependent variable and one or more independent variable by fitting a best-fit straight line to the data.

`predicts a continuous value (like salary, marks, price) using a linear relationship`

```python
import numpy as np
```

```python
X = np.array([1, 2, 3])
y = np.array([2, 4, 5])
```

## Hypothesis Function

The hypothesis function is the mathematical equation used by Linear Regression to predict the output value for a given input.

`model's prediction formula`

$$h_\theta(x) = \theta_0 + \theta_1x$$

 $$\theta_0 = \text{intercept/bias}$$
 $$\theta_1 = \text{slope}$$

```python
theta0 = 0 
theta1 = 0
```

```python
learning_rate = 0.1
iterations = 1000
m = len(X)
```

## Gradient Descent

optimization algorithm used to minimize the cost function by iteratively updating the model parameters in the direction of the steepest decrease cost.

$$\theta_j := \theta_j - \alpha \frac{\partial J}{\partial \theta_j}$$

```python
# Gradient Descent loop

for i in range(iterations):
    
    # hypothesis (predictions)
    predictions = theta0 + theta1 * X
    
    # errors
    errors = predictions - y
    
    # gradients
    d_theta0 = (1/m) * np.sum(errors)
    
    d_theta1 = (1/m) * np.sum(errors * X)
    
    
    # update parameters
    
    theta0 = theta0 - learning_rate * d_theta0
    
    theta1 = theta1 - learning_rate * d_theta1
    
```

```python
print("Theta0", theta0)
print("Theta1", theta1)
```

```python
# make predictions
hours = 4

predicted_score = theta0 + theta1 * hours

print("Predicted Score: ", predicted_score)
```

## Cost Function

mathematical function that measures how well or poorly a machine learning model performs by quantifying the difference between actual and predicted values.

`tells how wrong the model is`

$$J_\theta = \frac{1}{2m} \sum_{i=1}^{m} (h_\theta(x_i) - y_i)^2$$

$${m} = \text{no. of  training  examples}$$
$$h_\theta(x_i) = \text{prediction}$$
$$y_i =\text{actual value}$$



```python
cost = (1/ (2*m)) * np.sum(errors ** 2)
print(cost)
```
