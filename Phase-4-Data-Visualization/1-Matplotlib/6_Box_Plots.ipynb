# **Box Plots**

**A box plot is a statistical visualization that summarizes the distribution of a dataset using five key numbers:**

- Minimum (lowest non-outlier) 
- First Quartile (Q1) - 25th percentile
- Median (Q2) - 50th percentile
- Third Quartile (Q3) - 75th percentile
- Maximum (highest non-outlier)
---

### **When to Use a Box Plot**

Use a box plot when:

- We want to detect outliers  
- We want to compare distributions between multiple groups  
- We want to understand spread and skewness  
- We need a summary of data distribution without showing all data points  

---

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = [7, 8, 5, 6, 9, 4, 10, 12, 15] 

plt.boxplot(data) 
plt.ylabel("Values") 
plt.title("Simple Box Plot") 
plt.grid()
plt.show()
```

## Key Components

- **Box** → Represents the Interquartile Range (`IQR = Q3 - Q1`)  
- **Line inside box** → Median  
- **Whiskers** → Extend to minimum and maximum values (excluding outliers)  
    `Lower whisker = Q1 - 1.5 × IQR`

    `Upper whisker = Q3 + 1.5 × IQR`
- **Outliers** → Data points far from the rest (usually shown as dots)

---


## Box Plot Customizations (Matplotlib)

| Parameter      | Purpose                                         | Example                                    |
| -------------- | ----------------------------------------------- | ------------------------------------------ |
| `vert`         | Controls vertical or horizontal orientation     | `vert=False`                               |
| `patch_artist` | Fills box with color                            | `patch_artist=True`                        |
| `showmeans`    | Displays the mean value on the plot             | `showmeans=True`                           |
| `widths`       | Controls the width of the box                   | `widths=0.5`                               |
| `notch`        | Adds a notch to show median confidence interval | `notch=True`                               |


```python
plt.boxplot(
    data,
    vert=False,
    patch_artist=True,
    showmeans=True, 
    notch=True
)
```

| Parameter      | Purpose                                         | Example                                    |
| -------------- | ----------------------------------------------- | ------------------------------------------ |
| `boxprops`     | Customizes box properties (color, linewidth)    | `boxprops=dict(facecolor="lightblue")`     |
| `medianprops`  | Customizes median line                          | `medianprops=dict(color="red")`            |
| `whiskerprops` | Customizes whiskers                             | `whiskerprops=dict(color="black")`         |
| `capprops`     | Customizes caps at whisker ends                 | `capprops=dict(color="green")`             |
| `flierprops`   | Customizes outliers                             | `flierprops=dict(marker="o", color="red")` |
| `showfliers`   | Shows or hides outliers                         | `showfliers=False`          

```python
plt.boxplot(
    data,
    patch_artist=True,
    boxprops=dict(facecolor="lightpink"),
    medianprops=dict(color="magenta"),
    whiskerprops=dict(color="red"),
    capprops=dict(color="darkblue")
)
```

## Multiple Box Plots

```python
group1 = np.random.normal(50, 10, 100)
group2 = np.random.normal(60, 15, 100)

plt.boxplot([group1, group2], tick_labels=["Group1", "Group2"])
plt.grid(True) 
plt.show()
```
