## **`NumPy Vector Normalization`**

A quite common example of broadcasting in Vector Normalization. 

This is very 
common in machine learning and data preprocessing.  

Let’s take an example of Standard Vector Normalization i.e. transforming an array 
such that it has: 

-  `mean = 0` 
- `standard deviation = 1` 

For each element xi in vector,   ` xi ^ normalized = (xi − μ)/σ `
Where:  

- `μ` = mean of the vector (or column) 
- `σ`= standard deviation 

```python
import numpy as np
```

```python
# standard Vector Normalization
arr = np.array([[1, 2], [3, 4]]) 
mean = np.mean(arr) 
std_dev = np.std(arr) 
normalized_arr = (arr - mean) / std_dev 

print(normalized_arr)
```

```python
# column wise Normalization
arr = np.array([[1, 2], [3, 4], [5, 6]]) 
mean = np.mean(arr, axis=0) 
std_dev = np.std(arr, axis=0) 
norm = (arr - mean) / std_dev
print(norm)
```
