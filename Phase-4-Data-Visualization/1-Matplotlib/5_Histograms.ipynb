# **Histograms**

A **histogram** shows the distribution of numerical data by:

- Dividing values into **bins (ranges)**
- Counting how many values fall into each bin (**frequency**)

`bins` controls how the data is divided.

It can be:

- An **integer** → Number of equal-width bins  
- A **sequence** → Custom bin edges  


#### **When to Use a Histogram**

Use a histogram when:

- You want to show the distribution of continuous data  
- You need to analyze frequency, spread, skewness, or patterns  
- You are working with large datasets and want to summarize trends  


```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = np.random.randn(1000)

plt.hist(data, bins=30)
plt.title("Basic Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show
```

```python
np.random.seed(0)
data = np.random.normal(70, 10, 100)

plt.hist(data, bins=10, color='skyblue', edgecolor="black")
plt.title("Basic Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show
```

```python
dataval = [1, 2, 2, 3, 3, 3, 4, 4, 5, 5, 5, 5]

plt.hist(dataval, edgecolor="black", bins=[1, 2, 3, 4, 5 , 6]) # sequence
plt.title("Basic Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show
```

### **Histogram Customizations**

| Parameter | Purpose | Example |
|------------|----------|----------|
| `bins` | Controls the number of intervals (bars) | `bins=10` |
| `range` | Defines the lower and upper range of bins | `range=(0, 100)` |
| `color` | Sets the color of bars | `color="skyblue"` |
| `edgecolor` | Sets the border color of bars | `edgecolor="black"` |
| `linewidth` | Controls thickness of bar borders | `linewidth=1.5` |
| `alpha` | Controls transparency (0 to 1) | `alpha=0.7` |
| `density` | Normalizes histogram (shows probability instead of frequency) | `density=True` |
| `histtype` | Changes bar style (`bar`, `step`, `stepfilled`) | `histtype="step"` |
| `cumulative` | Shows cumulative histogram | `cumulative=True` |

```python
plt.hist(
    data,
    bins=30,
    color="darkblue",
    edgecolor="black",
    alpha=0.7,
    linewidth=1.5,
    histtype="bar",
    cumulative=True
)
```

```python
plt.hist(data, bins=30, density=True, edgecolor="black")
```

### **Multiple Datasets in Histogram**

```python
data1 = [1, 2, 2, 3, 3, 3, 4, 4, 5, 5, 5, 5]
data2 = [2, 3, 3, 4, 4, 5, 5, 5, 6]

plt.hist(data1, bins=5, alpha=0.5, label="Group A", color="red", edgecolor='black')
plt.hist(data2, bins=5, alpha=0.5, label="Group B", color="green", edgecolor='black')
plt.legend()
plt.show()
```

## **`axvline()` in Histogram**

`axvline` (Axix Vertical line) is used to draw a vertical line at a specific x-value in a plot.

```python
mean_value = np.mean(data)
plt.hist(data, bins=30, edgecolor="black")
plt.axvline(mean_value, color="red", linestyle="--")
plt.show()
```

```python
median = np.median(data)
mean = np.mean(data)

plt.hist(data, bins=30, edgecolor="black")
plt.axvline(mean, color="red", label="Mean")
plt.axvline(median, color="blue", label="Median")
plt.legend()
```
