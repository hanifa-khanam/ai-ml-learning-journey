### `requests` Library (The Downloader)

`requests` is a Python library used to send **HTTP requests**.

It is responsible for **downloading webpage content (HTML)** from the internet.

---

### What is HTTP?

**HTTP** = *HyperText Transfer Protocol*

It is the **rulebook for communication** between:

-  **Client** → (Your Python script / browser)
-  **Server** → (Website or API server)

HTTP defines how requests are sent and how responses are returned.


#### **Common HTTP Methods**

| Method | Meaning | Common Use | Simply | 
|--------|----------|------------| ------------ |
| `GET` | Retrieve data | Web scraping, fetching pages | Give me data  |
| `POST` | Send data | Submitting forms, sending data to server | Here is new data |
| `PUT` | Update data | Updating resources via APIs | Update existing data |
| `DELETE` | Delete data | Removing resources via APIs | Remove this data|

---


```python
import requests
```

```python
URL  = "https://www.scrapethissite.com/pages/simple/"

res = requests.get(URL) 
```

### Common `requests` Response Attributes

| Code | Description |
|------|------------|
| `res.status_code` | HTTP status code (`200` = OK) |
| `res.text` | HTML content of the page (as a string) |
| `res.content` | Raw bytes of the response |
| `res.headers` | Metadata (content type, server info, etc.) |

### BeautifulSoup — The HTML Parser

**BeautifulSoup** reads raw HTML and converts it into a **structured tree format**,  
making it easy to navigate and extract data.

It transforms messy webpage code into something we can search, filter, and analyze.

**Parsing** = Analyzing raw text and breaking it into a **structured format**.

In web scraping:

Raw HTML ➝ Parsed Tree ➝ Easy Data Extraction

`requests` ➝ Download HTML  
`BeautifulSoup` ➝ Parse HTML  


```python
from bs4 import BeautifulSoup
```

```python
with open("scraped_data/data.html", "r") as f:
    html_content = f.read() 
    
soup = BeautifulSoup(html_content, "lxml")
```

```python
soup.find("h3") 
soup.h3
```

```python
all_countries = [] 

all_h3 = soup.find_all("h3") 

for h3 in all_h3:
    name = h3.get_text(strip=True)
    #print(h3.find_parent("div")["class"])
    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    
    all_countries.append([name, population]) 
    
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(all_countries, columns=["Country Name", "Population"])
```
