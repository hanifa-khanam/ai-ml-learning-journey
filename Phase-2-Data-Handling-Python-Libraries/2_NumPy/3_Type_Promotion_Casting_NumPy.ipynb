## **`NumPy Type Promotion & Casting`**


**`Type promotion means:`**

- When you perform an `operation on arrays of different data types`, `NumPy automatically converts them to a common, safer type`.

```python
import numpy as np
```

```python
a = np.array([1, 2, 3], dtype=np.int32) 
b = np.array([1.5, 2.5, 3.5], dtype=np.float32) 

c = a + b
print(c) 
print(c.dtype)
```

**NumPy always promotes to the most precise data type to avoid data loss. So:**

**`int32 + float32 → float64`**

**NumPy chooses `float64` for maximum safety.**

| `Operation`       | `Result dtype` |
| --------------- | ------------ |
| bool + bool     | bool         |
| int + bool      | int          |
| int + float     | float        |
| float + complex | complex      |


```python
arrsum = np.array([1, 2]) + np.array([True, False])
print(arrsum)
print(arrsum.dtype)
```

```python
a1 = np.array([1, 2], dtype=np.int16) + np.array([1.5, 2.5])
print(a1) 
print(a1.dtype)
```
