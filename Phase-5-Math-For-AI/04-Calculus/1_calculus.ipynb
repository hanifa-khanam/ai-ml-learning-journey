# Compute Derviative Numerically

```python
import numpy as np
```

```python
def f(x):
    return x**2

def derivative(x, h=1e-5):
    return (f(x+h)-f(x))/h

x = 3 
print("Approx derivative at x=3: ", derivative(x))
```

# Find Minimum using derivative

```python
def f(x):
    return x**2 + 4*x + 4

x = -2
print("Minimum Value: ", f(x))
```

# Effect of Weight and Bias

```python
def model(x, w, b):
    return x*w + b

x = 10

print("Different Weights: ")
for w in [0.5, 1, 2]:
    print(model(x, w, 0))
    
print("\nDifferent bias: ") 
for b in [0, 5, 10]:
    print(model(x, 1, b))
```
