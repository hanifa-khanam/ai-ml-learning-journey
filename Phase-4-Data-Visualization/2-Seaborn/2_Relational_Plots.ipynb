# Relational Plots in Seaborn

`Purpose`: **Understand relationships between two numerical variables**

Common plots:

   - `scatterplot()`

   - `lineplot()`

```python
import seaborn as sns
import matplotlib.pyplot as plt 
```

# Scatter Plot

```python
df1 = sns.load_dataset("diamonds") 
df1
```

```python
sns.scatterplot(
    data=df1,
    x="carat",
    y="price",
    hue="cut"
)
plt.grid()
```

# Line Plot

```python
df2 = sns.load_dataset("flights") 
df2
```

```python
sns.lineplot(
    data=df2,
    x="year",
    y="passengers"
)
plt.grid(True)
```

# High-Level Relational Plot

`relplot()` is a **figure-level function** that combines relational plots.

```python
df = sns.load_dataset("tips") 

sns.relplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex"
)
```
