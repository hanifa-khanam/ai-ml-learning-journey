## **`Vectorization & Broadcasting in NumPy`**

---

#### **Vectorization**

Vectorization means performing `operations on entire arrays at once` without explicit 
Python loops.  

- NumPy uses `C-level implementations` internally → much faster than Python loops. 

- Makes code `shorter, cleaner, and faster`. 

```python
import numpy as np
```

```python
arr1 = np.array([1, 2, 3, 4, 5]) 

square_arr = arr1 ** 2     # Square all of nums 
print(square_arr) 

arr2 = np.array([6, 7, 8, 9, 10]) 
print(arr1 + arr2)
```

---

#### **Broadcasting**

Broadcasting allows NumPy to `perform operations on arrays of different shapes by automatically expanding the smaller one.`

It’s basically scaling arrays without using extra memory. 

- No need to manually reshape arrays. 

- Useful for `combining arrays of different dimensions`. 

**`Broadcasting Rules`**

- Align shapes from the `right to left`.

- Dimensions are compatible if:

            • They are `equal`, OR

            • One of them is `1`

            • `Missing` (smaller array can be “stretched”).

- If not compatible →  error

```python
# 1D array + scalar

arr = np.array([1, 2, 3]) 
print(arr + 10)
```

```python
# 2D + 1D 

arr2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

arr1d = np.array([10, 20, 30]) 

print(arr2d + arr1d)
```

```python
# 2D + scalar
arr2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(arr2d + 10)
```

```python
a = np.array([12, 24, 36, 45, 60, 72]) 
b = np.array(["Pass", "Fails"]) 

res = np.where(a > 35, b[0], b[1])
print(res)
```

```python
# matrix X vector
m = np.array([
    [1, 2],
    [3, 4]
])

v = [10, 20] 
res = m * v
print(res)
```
