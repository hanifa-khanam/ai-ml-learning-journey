# **Stack Plot**

**A stack plot is a type of plot where multiple data series are stacked on top of each other.**

Each layer shows the contribution of one category, and the top line shows the cumulative total.

---

#### **When to Use Stack Plot**

Use a stack plot when you want to:

- Show cumulative data over time  
- Compare multiple components  
- Highlight trends in parts and whole  

```python
import matplotlib.pyplot as plt
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'] 

product_A = [3, 4, 5, 6, 7, 8, 9]
product_B = [1, 2, 2, 3, 4, 5, 6]
product_C = [2, 2, 3, 3, 5, 5, 6] 

plt.stackplot(
    days,
    product_A, 
    product_B, 
    product_C, 
    labels=["Product A", "Product B", "Product C"]
)
plt.legend()
```

```python
colors = ["skyblue","pink","yellow"]

plt.stackplot(days, product_A, product_B, product_C, colors=colors, labels=["Product A", "Product B", "Product C"])
plt.grid()
plt.legend()
```

### Important Parameters of `stackplot()`

| Parameter  | Meaning                  | Example                               |
| ---------- | ------------------------ | ------------------------------------- |
| `labels`   | Labels for legend        | `labels=["Study","Sleep","Exercise"]` |
| `colors`   | Custom colors for areas  | `colors=["blue","green","orange"]`    |
| `alpha`    | Controls transparency    | `alpha=0.8`                           |
| `baseline` | Controls stacking method | `baseline="zero"`                     |


```python
days = [1, 2, 3, 4, 5, 6, 7]
study = [2, 3, 4, 3, 5, 4, 6]
sleep = [7, 6, 6, 7, 6, 6, 4] 
exercise = [1, 1, 1, 1, 1, 1, 1]

plt.stackplot(
    days, 
    study,
    sleep,
    exercise,
    labels=["study", "Sleep", "Exercise"],
    colors=["magenta", "red", "green"],
    alpha= 0.8,
    baseline="zero"
)

plt.title("Daily Activity Distribution") 
plt.xlabel("Days") 
plt.ylabel("Hours")
plt.legend() 
plt.grid() 
plt.show()
```
