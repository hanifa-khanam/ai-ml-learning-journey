## **`Pandas with CSV and JSON data`**

- `CSV files` are read using `pd.read_csv()` and `JSON files` using `pd.read_json()`. 

- Pandas automatically converts them into DataFrames and provides options to `handle separators, headers, missing values, and nested JSON structures`.

---

```python
import pandas as pd
```

### **1️⃣ — Working with CSV Files**

**`CSV = Comma Separated Values`**

```python
df = pd.read_csv("employee_data.csv")
print(df, type(df))
```

### **2️⃣ — Working with JSON Data**

**`JSON = JavaScript Object Notation`**

```python
df = pd.read_json("employee_data.json") 
print(df, type(df))
```
