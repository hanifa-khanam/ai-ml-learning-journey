# Raw Data Issues

The given dataset contains multiple data quality problems that must be fixed before analysis or machine learning:

- **Inconsistent rating formats**: `"5 "`, `"four"`, `"two"`, `"3.5"`
- **Extra whitespaces** in names and ratings
- **Mixed data types**: string numbers instead of numeric
- **Missing values**: age not present for some users
- **Duplicate users**: same name appearing multiple times
- **Unstructured feedback text**: case inconsistency


##  Data Cleaning Strategy (VERY IMPORTANT)

**What we will do:**

1. **Load JSON data safely**  
2. **Normalize ratings into numeric form**  
3. **Handle missing age values**  
4. **Remove duplicate users**  
5. **Keep only clean, structured records**  

This follows a real-world ETL pipeline:  

**Extract → Transform → Load**


````python
import json
````

###  Step 1: Loading the Data

###  Concept Explanation

**Why we use `json.load()` instead of `file.read()`?**

- `json.load()` converts JSON directly into Python objects  
- It avoids manual parsing  
- It is safer and cleaner


````python
# -----------------------------
# Function: load_data
# Purpose : Load JSON data from file
# -----------------------------
def load_data(filename):
    """
    Reads JSON data from a file and returns
    it as a Python list of dictionaries.
    """
    with open(filename, "r") as file:
        data = json.load(file)
    return data

````

````python
# Load data from JSON file
data = load_data("store_data.json")

# Preview raw data
for d in data:
    print(d)
````

### Step 2: Cleaning & Structuring the Data

##### Key Cleaning Operations

##### 1️⃣ Rating Normalization

Ratings appear as:  

- Text (`"four"`, `"two"`)  
- Numbers with spaces (`"5 "`)  
- Decimal strings (`"3.5"`)  

We convert text numbers into integers using a **mapping dictionary**.

##### 2️⃣ Handling Missing Age

- Some users do not have an `"age"` key  
- Missing values are explicitly set to `None`  
- This avoids runtime errors in analysis

##### 3️⃣ Deduplication

- Duplicate users exist (same name repeated)  
- Only the **first occurrence** is kept  
- A `set()` is used for **efficient duplicate detection**


````python
# -----------------------------
# Function: clean_data
# Purpose : Clean and structure raw data
# -----------------------------
def clean_data(data):
    """
    Cleans raw user data by:
    - Converting textual ratings to numbers
    - Handling missing age values
    - Removing duplicate users
    """
    
    # Mapping text ratings to numeric values
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}    
    
    
    cleaned_data = []    # Stores final cleaned records
    unique_users = set()  # Tracks unique user names
    
    
   
    for user in data:     # Looping through data
        # -----------------------------
        # Clean rating
        # -----------------------------
        raw_rating = str(user["rating"]).strip().lower() 
         
        if raw_rating in text_to_num:         
            user["rating"] = text_to_num[raw_rating]     
    
    
        # -----------------------------
        # Handle missing age
        # -----------------------------
        raw_age = user.get("age")
        if raw_age == None:
            user["age"] = None


        # -----------------------------
        # Clean and deduplicate name
        # -----------------------------
        if(user["name"].strip() in unique_users):
            continue
        
        unique_users.add(user["name"])
        cleaned_data.append(user)
    
    return cleaned_data





    
        
````

````python
data = clean_data(data) 

for i, record in enumerate(data):
    print(f"{i+1} : {record}")
    print()
    
    
    
````

### Step 3: Getting Meaningful insights from data
- What's Average rating?
- What % of users gave poor rating (<3)?

````python
def get_insights(data):
    
    #ave rating
    total_rating = 0
    
    for user in data:
        total_rating += float(user["rating"]) 
    
    print(f"Average rating = {total_rating/len(data)}") 
    
    # percentage of users with poor rating
    poor_ratings = 0
    
    for user in data:
        if(float(user["rating"]) < 3):
            poor_ratings += 1
    
    print(f"% of users with poor rating = {poor_ratings/len(data) * 100}%")
    
````

````python
get_insights(data)
````

### Step 4: Build a Product recommendation feature (rule based, not ML)
- If user rating >= 4, recommend same brand products.
- If user rating < 4, recommend different brand products.

````python
# Build the recommendation feature
def get_recommendation(data):
    recommendations = []

    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]

        if(float(user["rating"]) >= 4):
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        recommendations.append(curr_recomm)
    return recommendations
    
````

````python
get_recommendation(data)
````
