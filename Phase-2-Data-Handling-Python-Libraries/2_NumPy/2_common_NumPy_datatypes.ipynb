## **`Common NumPy Data Types`**


| Category               | dtype        | Memory Size | Description                               | Example                                      | Common Use Case                       |
| ---------------------- | ------------ | ----------- | ----------------------------------------- | -------------------------------------------- | ------------------------------------- |
| **Integer (signed)**   | `int8`       | 1 byte      | Small signed integer (-128 to 127)        | `np.array([1,2], dtype=np.int8)`             | Image processing, small counts        |
|                        | `int16`      | 2 bytes     | Medium signed integer (-32,768 to 32,767) | `np.array([1000,2000], dtype=np.int16)`      | Sensor data, compact storage          |
|                        | `int32`      | 4 bytes     | Standard signed integer                   | `np.array([10**6, 2*10**6], dtype=np.int32)` | ML features, scientific computing     |
|                        | `int64`      | 8 bytes     | Large signed integer                      | `np.array([10**12], dtype=np.int64)`         | Big data, finance                     |
| **Integer (unsigned)** | `uint8`      | 1 byte      | 0 to 255                                  | `np.array([0, 255], dtype=np.uint8)`         | Images (grayscale, RGB channels)      |
|                        | `uint16`     | 2 bytes     | 0 to 65,535                               | `np.array([1000,50000], dtype=np.uint16)`    | High-res images, counts               |
|                        | `uint32`     | 4 bytes     | 0 to 4.2 billion                          | `np.array([1_000_000], dtype=np.uint32)`     | Large positive datasets               |
|                        | `uint64`     | 8 bytes     | Very large positive integers              | `np.array([10**12], dtype=np.uint64)`        | Finance, scientific datasets          |
| **Float**              | `float16`    | 2 bytes     | Half-precision float                      | `np.array([1.5, 2.3], dtype=np.float16)`     | Memory-efficient ML, low-precision DL |
|                        | `float32`    | 4 bytes     | Single-precision float                    | `np.array([1.5, 2.3], dtype=np.float32)`     | ML features, GPU-friendly             |
|                        | `float64`    | 8 bytes     | Double-precision float                    | `np.array([1.5, 2.3], dtype=np.float64)`     | High-precision scientific computing   |
| **Complex**            | `complex64`  | 8 bytes     | 32-bit real + 32-bit imaginary            | `np.array([1+2j], dtype=np.complex64)`       | Signal processing, physics            |
|                        | `complex128` | 16 bytes    | 64-bit real + 64-bit imaginary            | `np.array([1+2j], dtype=np.complex128)`      | Scientific simulations                |
| **Boolean**            | `bool_`      | 1 byte      | True/False                                | `np.array([True, False], dtype=np.bool_)`    | Masks, filters, indexing              |
| **String / Unicode**   | `str_`       | Variable    | Fixed-length string                       | `np.array(['a', 'b'], dtype=np.str_)`        | Text datasets, labels                 |
| **Object**             | `object_`    | Variable    | Stores any Python object                  | `np.array([1, "a"], dtype=np.object_)`       | Mixed types, custom objects           |


```python
import numpy as np
```

```python
arr1 = np.array([1, 2, 3, 4, 5], dtype="int16")
print("Memory:", arr1.nbytes)


arr2 = np.array([1, 2, 3, 4, 5], dtype="float32")
print("Memory:", arr2.nbytes)

arr = np.array([1,2,3,0], dtype=np.bool_)
print(arr)
print("Memory:", arr.nbytes)


```

**Memory Tips**

- `Memory matters: Smaller dtypes → less memory → faster computation.`

- `Default dtype: If you don’t specify, integers → int64, floats → float64.`

- `Type conversion: Use .astype() to convert between dtypes.`

- `GPU/ML tip: float32 is preferred in deep learning because GPUs work faster with it.`

```python
# Create a NumPy array of 1 million integers using int32 and int64. Check memory difference using .nbytes

size = 1_000_000
py_list = list(range(size))
arr1 = np.array(py_list, dtype=np.int32) 
print(f"Memory Size of int32 array: {arr1.nbytes}")

arr2 = np.array(py_list, dtype=np.int64) 
print(f"Memory Size of int64 array: {arr2.nbytes}")

print(f"Memory Difference = {arr2.nbytes - arr1.nbytes}")
```

- *`Deep learning models` prefer `float32` because it `halves memory` and `runs faster on GPUs`.*

-That’s why `TensorFlow & PyTorch` default to `float32`

```python
# Convert a float64 array of [1.1, 2.2, 3.3] to float32 and compare memory.
arr64 = np.array([1.1, 2.2, 3.3], dtype=np.float64)
arr32 = arr64.astype(np.float32) 

print(f"float64 memory: {arr64.nbytes} bytes. ")
print(f"float32 memory: {arr32.nbytes} bytes. ")
```

`Boolean masks are used in:`

- Data cleaning

- Removing outliers

- Selecting rows in ML datasets

- Feature engineering

- This single concept powers Pandas filtering, ML preprocessing, and even computer vision pipelines.

```python
# Create a boolean mask for numbers > 10 in [5, 12, 18, 7].

arr = np.array([5, 12, 18, 7]) 

mask = arr > 10
print(mask)

# applying mask
filtered = arr[mask] 
print(filtered)
```

```python
# Common Data Types  
arr = np.array([1, 2, 3, 4, 5])  
arr2 = np.array([1.0, 2.0, 3.0])  
arr3 = np.array(["hello", "world", "prime", "ai/ml"])  

print(arr.dtype) # int64 
print(arr2.dtype) # float64  
print(arr3.dtype) # U  

# Complex Numbers  
arr1 = np.array([2 + 3j])  
arr2 = np.array([5 + 8j])  

print(arr1, arr1.dtype)  
print("sum of complex arr1 + arr2: ", arr1 + arr2)  
print("subtraction of complex arr2 - arr1: ", arr2 - arr1)  

# Objects  
arr = np.array (["hello", {1, 2, 3}, 3.14])  
print(arr, arr.dtype) 
```
