# Categorical Plots

Categorical plots are used to visualize data where one variable is categorical (discrete) and the other is usually numerical.


#### Main Categorical Plots in Seaborn

| Plot           | Purpose                                    |
| -------------- | ------------------------------------------ |
| `countplot()`  | Counts number of observations per category |
| `barplot()`    | Shows average value for categories         |
| `boxplot()`    | Shows distribution and outliers            |
| `violinplot()` | Shows detailed distribution                |



```python
import seaborn as sns
import matplotlib.pyplot as plt
```

## 1. Count Plot

Used to count occurences of categories.

```python
df = sns.load_dataset("tips") 

df.head() 
```

```python
sns.countplot(data=df, x="day") 
plt.grid()
plt.show()
```

## 2. Bar Plot

Shows average values across categories.

```python
sns.barplot(data=df, x="day", y="tip", hue="sex") 

plt.grid()
plt.show()
```

## 3. Box Plot

Used to visualize data distribution and outliers.

```python
sns.boxplot(data=df, x="day", y="tip")
plt.grid()
```

## 4. Violin Plot

Shows distribution shape for categories.

```python
sns.violinplot(data=df, x="day", y="total_bill")

plt.show()
```
