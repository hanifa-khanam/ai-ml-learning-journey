# Matrix Plots

Matrix Plots are used to visualize 2D data arranged in a matrix or table, such as correlation matrices, or pivot tables

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

## 1. Heatmap

A heatmap is a graphical representation of data where valuses are shown using colors.

### Key Use Cases in Machine Learning (ML)
* **Feature Selection:** Identify features that have a strong relationship with the target variable.
* **Detect Multicollinearity:** Spot independent variables that are highly correlated with each other.
* **Improve Model Performance:** Helps in deciding which features to keep or drop.
* **Remove Redundant Features:** Streamlines the dataset by removing features that provide overlapping information.



```python
tips = sns.load_dataset("tips")
corr = tips.corr(numeric_only=True)  # correlation between numerical columns 

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

```

## 2. Correlation Theory
Correlation measures the statistical relationship between two numerical variables.

### The Scale
* **+1.0**: Strong **Positive** correlation ( both variablesincrease together).
* **0.0**: **No** correlation (No relationship between variables).
* **-1.0**: Strong **Negative** correlation (one increased, other decreases).

### Correlation Strength Reference
| Value Range | Meaning |
|:---|:---|
| **0.7 to 1.0** | Strong correlation |
| **0.3 to 0.7** | Moderate correlation |
| **0.0 to 0.3** | Weak correlation |

> **Note:** Correlation works **only on numerical data**. Categorical data must be removed or encoded (e.g., One-Hot Encoding) before calculating correlation.

### Important Heatmap Parameters
When using `sns.heatmap()`, these parameters enhance readability:
* `annot=True`: Displays the data values in each cell.
* `cmap="coolwarm"`: Sets the color gradient (Red = Positive, Blue = Negative).
* `linewidths`: Adds a small gap between cells to make them distinct.
* `fmt=".2f"`: Rounds the annotated values to 2 decimal places.

```python
flights = sns.load_dataset("flights") 
flights_pivot_data = flights.pivot(index="month", columns="year", values="passengers")
flights_pivot_data
```

```python
sns.heatmap(flights_pivot_data,
            cmap="coolwarm",
            annot=True,
            fmt="d")

plt.title("Passengers Heatmap")
plt.tight_layout()
```
