# Probability for AI/ML

This notebook builds practical intuition for probability using Python.

Topics covered:
- Basic probability and sample space
- Conditional probability
- Bayes' theorem
- Counting (permutations/combinations)
- Simulation (Law of Large Numbers)
- Basic distributions with random experiments

```python
import random
from math import comb

random.seed(42)
print("Setup complete. Ready to explore probability.")
```

```python
# 1) Basic probability

def probability(favorable_outcomes, total_outcomes):
    if total_outcomes == 0:
        raise ValueError("total_outcomes cannot be 0")
    return favorable_outcomes / total_outcomes

# Example: rolling an even number on a fair die
favorable = 3  # {2, 4, 6}
total = 6
print("P(Even on die) =", probability(favorable, total))
```

```python
# 2) Conditional probability
# P(A|B) = P(A and B) / P(B)

def conditional_probability(p_a_and_b, p_b):
    if p_b == 0:
        raise ValueError("p_b cannot be 0")
    return p_a_and_b / p_b

# Example: From a deck, probability of drawing a King given the card is a face card
# Face cards = J, Q, K => 12 cards; Kings among them = 4
p_king_given_face = conditional_probability(4/52, 12/52)
print("P(King | Face card) =", p_king_given_face)
```

```python
# 3) Bayes' theorem
# P(A|B) = (P(B|A) * P(A)) / P(B)

def bayes_theorem(p_b_given_a, p_a, p_b):
    if p_b == 0:
        raise ValueError("p_b cannot be 0")
    return (p_b_given_a * p_a) / p_b

# Example (medical test):
# P(Disease)=0.01, P(Positive|Disease)=0.95, P(Positive)=0.058
p_disease_given_positive = bayes_theorem(0.95, 0.01, 0.058)
print("P(Disease | Positive) =", round(p_disease_given_positive, 4))
```

```python
# 4) Counting techniques (combinations)
# Number of ways to choose r items from n items: nCr

def ncr(n, r):
    return comb(n, r)

# Example: choose 2 students from 10
print("10C2 =", ncr(10, 2))

# Example probability using combinations:
# Probability of getting exactly 2 heads in 3 fair coin tosses
ways_exactly_2_heads = ncr(3, 2)
total_outcomes = 2 ** 3
print("P(exactly 2 heads in 3 tosses) =", ways_exactly_2_heads / total_outcomes)
```

```python
# 5) Simulation: Law of Large Numbers (coin toss)

def simulate_coin_toss_probability(num_trials=10000):
    heads = 0
    for _ in range(num_trials):
        toss = random.choice(["H", "T"])
        if toss == "H":
            heads += 1
    return heads / num_trials

for trials in [10, 100, 1000, 10000]:
    estimate = simulate_coin_toss_probability(trials)
    print(f"Trials={trials:5d} -> Estimated P(Heads)={estimate:.4f}")

print("Expected theoretical value is 0.5")
```
