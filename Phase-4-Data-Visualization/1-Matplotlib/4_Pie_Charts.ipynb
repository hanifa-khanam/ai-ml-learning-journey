# **Pie Charts**

**A pie chart shows proportion of categories as slices of a circle.**

Each slice represents:  `Category Value / Total`

So pie charts answer:  **“What percentage does each category contribute?”**

**When to use?**

Use Pie Chart when:

 - We want to display part-to-whole relationships
 - we have categorical data
 - Data has few catagories (4-6); too many slices it hard to read

```python
import matplotlib.pyplot as plt
```

```python
labels = ['Sleep', 'Work', 'Commute', 'Exercise', 'Leisure', 'Chores']
sizes = [8, 7.5, 1.5, 1, 4, 2]

plt.pie(sizes, labels=labels)
```

###  Pie Chart Customizations

| Parameter | Purpose | Example |
|------------|----------|----------|
| `autopct` | Shows percentage on each slice | `autopct="%1.1f%%"` |
| `colors` | Adds custom colors to slices | `colors=["red","blue"]` |
| `wedgeprops` | Customizes slice properties (edgecolor, linewidth, linestyle, alpha, fill) | `wedgeprops={"edgecolor":"black"}` |
| `explode` | Highlights a slice by pulling it outward | `explode=[0.1, 0, 0]` |
| `startangle` | Rotates the chart | `startangle=90` |
| `shadow` | Adds shadow / 3D effect | `shadow=True` |

```python
# Dataset: Smartphone Market Share
brands = ['Samsung', 'Apple', 'Xiaomi', 'OPPO', 'Vivo', 'Others']
market_share = [20, 17, 12, 9, 8, 34]

plt.pie(market_share, labels=brands,
        autopct="%1.1f%%", 
        wedgeprops={'edgecolor': 'black', 'linewidth': 2, 'linestyle':''},
        explode=[0, 0, 0, 0, 0, 0.1],
        startangle=90, 
        shadow=True)
```
