# DBSCAN  Clustering

```python
from sklearn.datasets import load_iris
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris() 

X = iris.data
y = iris.target
```

```python
scaler = StandardScaler() 
X_scaled = scaler.fit_transform(X)
```

```python
sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 2])
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps = 0.8,
    min_samples= 5
)
```

```python
labels = dbscan.fit_predict(X_scaled)
```

```python
sns.scatterplot(x=X_scaled[:, 0], y=X_scaled[:, 2], c=labels)
```

##  DBSCAN  on linear-dataset

```python
from sklearn.datasets import make_moons

X2, y = make_moons(
    n_samples=300, 
    noise=0.05, 
    random_state=42
)

X2_scaled = scaler.fit_transform(X2)
```

```python
sns.scatterplot(x=X2_scaled[:, 0], y=X2_scaled[:, 1])
```

```python
# applying K-Means
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init="auto"
)

labels2 = kmeans.fit_predict(X2_scaled)
```

```python
sns.scatterplot(x=X2_scaled[:, 0], y=X2_scaled[:, 1], c=labels2)
```

```python
# applying dbscan

dbscan2 = DBSCAN(
    eps=0.5,
    min_samples=5
)

labels3 = dbscan2.fit_predict(X2_scaled)
```

```python
sns.scatterplot(x=X2_scaled[:, 0], y=X2_scaled[:, 1], c=labels3)
```
