# **Scatter Plots**

**Use a scatter plot when:**
 - you want to analyze the relationship between two variables
 - Data is numerical and continous
 - we don't want to imply a trend by connecting points

```python
import matplotlib.pyplot as plt
```

```python
temperature = [20, 22, 25, 27, 30, 32, 35, 37, 40]
ice_cream_sales = [100, 120, 150, 170, 200, 220, 250, 270, 300]

plt.scatter(temperature, ice_cream_sales) 
plt.title("Temperature Vs Ice Cream Sales") 
plt.xlabel("Temperature") 
plt.ylabel("Sales") 

plt.grid()
```

###  Scatter Plot Customizations

| Feature | Purpose | Example Parameter |
|----------|----------|------------------|
| **Color** | Changes the color of points | `color="blue"` or `c="red"` |
| **Size** | Controls the size of markers | `s=100` |
| **Marker** | Changes the marker shape | `marker="o"` |
| **Transparency** | Adjusts visibility level | `alpha=0.5` |
| **Colormap** | Applies gradient coloring based on values | `cmap="viridis"` |

```python
age = [22, 28, 34, 37, 42, 47, 53, 59, 63, 68, 72]
bp = [108, 112, 120, 118, 132, 138, 142, 148, 151, 158, 165]

colors = ["green" if x < 130 else "red" for x in bp]

plt.scatter(age, bp, c=colors, s=50, marker="o", alpha=0.6) 
plt.title("Age vs BP") 
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
```

```python
age = [22, 28, 34, 37, 42, 47, 53, 59, 63, 68, 72]
bp = [108, 112, 120, 118, 132, 138, 142, 148, 151, 158, 165]

colors = ["green" if x < 130 else "red" for x in bp]

plt.scatter(age, bp, s=bp, cmap="viridis", c=bp) 
plt.title("Age vs BP") 
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
plt.colorbar(label="BP")
```

### Adding Annotations

**Annotations are used to add explanatory text or markers to specific points in a plot**

```python
people = ['Person A','Person B', 'Person C', 'Person D', 'Person E', 'Person F', 'Person G', 'Person H', 'Person I', 'Person J', 'Person K']
age = [22, 28, 34, 37, 42, 47, 53, 59, 63, 68, 72]
bp = [108, 112, 120, 118, 132, 138, 142, 148, 151, 158, 165]

colors = ["green" if x < 130 else "red" for x in bp]

plt.scatter(age, bp, s=bp, cmap="viridis", c=bp) 
plt.title("Age vs BP") 
plt.xlabel("Age")
plt.ylabel("Blood Pressure")

plt.grid()
plt.colorbar(label="BP")

for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]), xytext=(age[i]+1, bp[i]+1))

plt.xlim(min(age), max(age)+10)
plt.ylim(min(bp), max(bp)+5)
```

- ### **Multiple Datasets on Scatter Plot**

**Used for:**

 - class comparison
 - cluster comparison



```python
cities = ["Karachi", "Lahore", "Islamabad", "Quetta", "Peshawar"]

# winter
winter_temp = [19, 12, 10, 5, 8]        # in °C
winter_humidity = [55, 65, 70, 40, 50] # in %

# summer
summer_temp = [35, 40, 38, 33, 42]     # in °C
summer_humidity = [75, 60, 55, 30, 45] # in %
```

```python
plt.scatter(winter_temp, winter_humidity, label="Winter") 
plt.scatter(summer_temp, summer_humidity, label="Summer")

plt.title("Summer Vs Winter Data")
plt.xlabel("temperature (C)")
plt.ylabel("Humidity (%)")
plt.legend()

```
