# Distribution Plots

#### **What Questions Do Distribution Plots Answer?**
* How is the data spread?
* Is it normal (Gaussian) or skewed?
* Are there outliers?
* Do we need transformation?

#### **Why is Data Distribution Critical in Machine Learning?**
Many ML models make the following assumptions about your data:
* Normal distribution
* No extreme skewness
* No heavy outliers

**If we ignore these assumptions, it can lead to:**
* Poor model accuracy
* Biased predictions
* Wrong feature importance

#### **Main Distribution Plots in Seaborn**

| Plot | Purpose |
| :--- | :--- |
| `histplot()` | Shows frequency distribution |
| `kdeplot()` | Shows smooth probability curve |
| `displot()` | Figure-level distribution plot |

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

# 1- Histogram (histplot())

```python
df = sns.load_dataset("tips") 

sns.histplot(data=df, x="total_bill", kde=True)
plt.show()
```

# 2- KDE Plot (kdeplot())

Shows a smooth probability density curve.

**Use case:**

- Understand shape of distribution
- Compare multiple distributions

```python
sns.kdeplot(data=df, x="total_bill")
```

# 3- Displot (displot()) 

Figure-level version (like relplot()) 

**Why use it?**

- Supports multiple plots
- Better layout control

Used in advanced EDA notebooks.

```python
sns.displot(data=df, x="total_bill", kde=True)
```
