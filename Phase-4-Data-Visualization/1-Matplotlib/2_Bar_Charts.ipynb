# **Bar Charts**


A bar plot (or bar chart) is a graph that displays categorical data using rectangular bars.

**When to use?**

we use bar chart when:

 - We are comparing different categories

 - The data is discrete, not continous

 - We want clear visual comparison

```python
import matplotlib.pyplot as plt
```

```python
categories = ["A", "B", "C"]
values = [100, 150, 80] 

plt.bar(categories, values) 
plt.title("Category Comparison") 
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

#### We can use `plt.text()` to add text annotations to the bars in the bar chart.

```python
plt.bar(categories, values) 
plt.title("Category Comparison") 
plt.xlabel("Categories")
plt.ylabel("Values")

for i, v in enumerate(values):
    plt.text(i, v, str(v), ha="center", va="bottom")
plt.show()    
```

- ### **Important Parameters in bar()**

| Parameter | Meaning           |
| --------- | ----------------- |
| color     | Bar color         |
| width     | Thickness of bars |
| edgecolor | Border color      |
| alpha     | Transparency      |


```python
plt.bar(categories, values, color="Green", width=0.5, edgecolor="black")
```

- ### **Multiple Datasets on Bar Chart**

**Used for:**

 - Region comparison

 - Gender comparison

 - Model comparison

```python
import numpy as np

x = np.arange(len(categories))
width = 0.35

values1 = [100, 150, 80] 
values2 = [180, 70, 280] 

plt.bar(x-width/2, values1, width, label="2024")
plt.bar(x+width/2, values2, width, label="2025")

plt.title("2024 Vs 2025 Subscription")
plt.xlabel("Categorie")
plt.ylabel("Subscription (in M)")
plt.xticks(x, categories)
plt.legend()
plt.show()
```

# **Horizontal Bar Charts**

**When to use?**

 - When category names are long
 - When ranking matters
 - When readability is priority

**Used for:**

 - Top 10 products
 - Feature importance (very common in ML!)

```python
languages = ["Python", "C++", "Java", "JavaScript", "Go", "Rust"] 
users = [15, 8, 12, 14, 5, 4]

plt.barh(languages, users, color="orange", edgecolor="red") 
plt.title("Programming Language Popularity")
plt.xlabel("Users (in M)")
plt.ylabel("Languages") 
```
