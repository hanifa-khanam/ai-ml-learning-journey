## **`Basic Visualization with Pandas`**

**Pandas has **built-in plotting** (on top of Matplotlib).**

> Why use Pandas for visualization?

- Very little code
- Quick insights
- Perfect for EDA (Exploratory Data Analysis)
- No need to write full Matplotlib code

---

```python
import pandas as pd
```

### **1️⃣ Histogram — `hist()`**

**A histogram shows the distribution of numerical data.**

It answers:

- How are values spread?
- Are most values small or large?
- Is the data skewed?

**Basic Syntax:**

df['column'].hist()


```python
df = pd.read_csv("employee_data.csv") 
df
```

```python
df["Age"].hist()
```

```python
df[["Age", "Salary"]].hist()
```

---

### **2️⃣ Line / Bar / Scatter — `plot()`**

**`plot()` is a **general-purpose plotting function** in Pandas.**

It answers:

- How are values spread?
- Are most values small or large?
- Is the data skewed?

**Basic Syntax:**

df.plot(kind='line')

**Common `kind` Values**
| kind        | Use Case            |
| ----------- | ------------------- |
| `'line'`    | Trends over time    |
| `'bar'`     | Category comparison |
| `'hist'`    | Distribution        |
| `'scatter'` | Relationship        |



```python
df
```

```python
df.plot(kind="scatter", x="Age", y="Salary")
```

```python
df.groupby("Department")["Salary"].mean().plot(kind="bar")
```

| Your Question                 | Data Type           | Plot Type |
| ----------------------------- | ------------------- | --------- |
| How is data distributed?      | Numeric             | hist      |
| Which category is bigger?     | Category vs Numeric | bar       |
| How does it change over time? | Time vs Numeric     | line      |
| Are two variables related?    | Numeric vs Numeric  | scatter   |


- **Golden Rule**

    - One variable → histogram
    - Two variables → line / bar / scatter
    - Time involved → line
