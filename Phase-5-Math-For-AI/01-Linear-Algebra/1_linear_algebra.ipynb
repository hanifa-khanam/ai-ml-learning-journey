# Project 1: Vector Operations 

What are we actually doing?

We are answering:

“How similar are two students?”

```python
import numpy as np
```

```python
# Two students (vectors)  [Math, English, coding]
student1 = np.array([80, 70, 90])
student2 = np.array([60, 65, 75])

# addition
print("Addition: ", student1 + student2)

# magnitude
print("Magnitude student1: ", np.linalg.norm(student1))

# Distance (Difference in each subject)
distance = np.linalg.norm(student1 - student2)

print("Distance: ", distance)  # small distance → similar, large distance → different
```

# Project 2: Similarity Engine (Dot Product)

Higher dot product = better recommendation


```python
# user preferences
user = np.array([5, 3, 0]) # action, comedy, romance

# movies
movie1 = np.array([4, 2, 0])
movie2 = np.array([1, 5, 4])

# Dot product similarity
score1 = np.dot(user, movie1)
score2 = np.dot(user, movie2)

print("Movie1 Score: ", score1)
print("Movie2 Score: ", score2)

if score1 > score2:
    print("Recommend Movie1")
else:
    print("Recommend Movie2")
```

# Project 3: ML Prediction using matrices

```python
# dataset (x)
# [size, bedrooms]
X = np.array([
    [1200, 3], # house 1
    [1500, 4], # house 2
    [800, 2]   # house 3
])

# weights (w)
W = np.array([
    [200], # weights for size 
    [50000] # weight for bedrooms
])

# Prediction 
y = np.dot(X, W)

print("Predicted Prices:\n", y)
```

# Project 4: Eigen + PCA

Which features actually matters?

```python
# sample data
X = np.array([
    [2, 3], 
    [3, 4],
    [4, 5]
])

# covariance matrix (How features move together)
cov_matrix = np.cov(X.T)

# Eigen decomposition
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("Eigenvalues: ", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```
