###  Importing Required Library

- `json` module is used to work with JSON files.
- JSON is a common format for storing and exchanging data.
- We import it before loading external data files.


```python
import json
```

###  Loading Raw Data

- This function reads data from a JSON file.
- It opens the file in read mode.
- `json.load()` converts JSON data into Python objects.
- Returning data makes the function reusable.


```python
def load_data(filename):
    with open (filename, "r") as file:
        data = json.load(file)
    return data
```

###  Inspecting Raw Data

- Printing raw data helps understand its structure.
- This step reveals missing values, inconsistencies, and duplicates.
- Always inspect data before cleaning or processing it.


```python
data = load_data("order_data.json")
for user in data:
    print(user)
    print()
```

###  Cleaning and Structuring the Data

- This function cleans messy real-world order data.
- It handles missing values, incorrect formats, and inconsistencies.
- The goal is to make data ready for analysis or machine learning.


#### 1: Cleaning Customer Names

- Extra spaces are removed using `.strip()`.
- Empty or missing names are replaced with `"Unknown"`.
- This prevents errors and ensures consistency.

#### 2: Cleaning Amount Field

- Text-based numbers are converted using a mapping dictionary.
- Numeric strings are converted into integers.
- Invalid or missing values are set to `None`.
- `isinstance()` ensures safe type checking.

#### 3: Standardizing Order Status

- Extra spaces are removed.
- Status values are converted to lowercase.
- This ensures consistency for filtering and analysis.


#### 4: Cleaning City Names

- Missing city values are replaced with `"Unknown"`.
- City names are formatted using title case.
- This improves readability and consistency.


#### 5: Removing Duplicate Records

- Duplicate orders are identified using `order_id`.
- A set is used to track already processed order IDs.
- Only unique orders are kept in the final dataset.


```python
def clean_data(data):
    text_to_num = {"five thousand": 5000} 
    
    cleaned_data = []
    unique_customers = set()
    
    for user in data:
        # handling customer name
        raw_customer = user.get("customer")
        
        if not raw_customer  or raw_customer.strip() == "":
            user["customer"] = "Unknown" 
        else:
            user["customer"] = user["customer"].strip()
            
            
        # handling amount
        raw_amount = user.get("amount") 
        
        if isinstance(raw_amount, str):
            raw_amount = raw_amount.strip().lower()
            
            if raw_amount in text_to_num:
                user["amount"] = text_to_num[raw_amount] 
            elif raw_amount.isdigit():
                user["amount"] = int(raw_amount) 
            else:
                user["amount"] = None
        else:
            user["amount"] = None
    
        # handling status
        raw_status = user.get("status") 
        
        if not raw_status:
            user["status"] = "Unknown" 
        else:
            user["status"] = user["status"].strip().lower()
            
        
        # handling city
        raw_city = user.get("city")
        
        if not raw_city:
            user["city"] = "Unknown" 
        else:
            user["city"] = user["city"].strip().title()
            
        # remove duplicates
        if user["order_id"] in unique_customers:
            continue
        
        unique_customers.add(user["order_id"]) 
        cleaned_data.append(user) 
        
    return cleaned_data
            
            
```

###  Final Cleaned Dataset

- Cleaned data is printed for verification.
- Each record is structured and consistent.
- This dataset is now ready for analysis or ML pipelines.


```python
data = clean_data(data)

for index, customer in enumerate(data):
    print(f"{index+1} :  {customer}")
    
```
