## **`Reshaping Data in Pandas: melt() & pivot()`**

>**Same data, different shape → different insight**

Sometimes our data is:

 - Good for humans to read
 - Bad for analysis or plotting

So we reshape it.

`Wide → many columns`
`Long → fewer columns, more rows`

---

```python
import pandas as pd
```

### **1️⃣ `melt()` — Wide ➜ Long**

`melt()` **turns columns into rows**

**Why do we need melt()?**

- Visualization (Seaborn loves long data)
- EDA
- Feature engineering
- Standardized datasets


```python
"""syntax: 

pd.melt(
    df,
    id_vars= [fixed_columns],
    value_vars= [columns_to_melt],
    var_name= new_column_name,
    value_name= new_value_name
)
"""
df = pd.DataFrame({
    "Month" : ["Jan", "Feb", "Mar"], 
    "Product_A" : [120, 150, 170],
    "Product_B" : [90, 110, 130],
    "Product_C" : [60, 80, 100]
})

df
```

```python
melted_data = df.melt(
    id_vars=["Month"],
    value_vars=["Product_A", "Product_B", "Product_C"],
    var_name="Product_Categ",
    value_name="Sales"
)

melted_data
```

---

### **2️⃣ `pivot()` — Long ➜ Wide**

`pivot()` **turns rows into columns.**

**Why do we need pivot()?**

- Reports & dashboards
- Feature matrix for ML
- Readable tables

> Important Rule of pivot():  Index + columns combination must be unique

```python
"""Syntax:
df.pivot(
    index='row_identifier',
    columns='column_identifier',
    values='cell_values'
)
"""

melted_data
```

```python
# converting melted_data to original
pivot_data = melted_data.pivot(
    index="Month",
    columns="Product_Categ",
    values="Sales"
)
pivot_data
```

---

### **2️⃣ `pivot_table()`**

`pivot_table()` is used to **summarize and reshape data** when **duplicate values exist.**

> It is like `pivot() + groupby() + aggregation` in one step.

**Why do we need pivot_table()?**

- `pivot()` fails when there are **duplicate combinations**
- Real-world data often has:
    - Multiple transactions
    - Repeated records

- `pivot_table()` **solves this by aggregating duplicates**

> Industry data → pivot_table() is safer than pivot()

```python
"""Syntax:
pd.pivot_table(
    data,
    index=...,     # rows
    columns=...,   # columns
    values=...,    # numbers to summarize
    aggfunc=...    # how to combine duplicates
)
"""
data = pd.DataFrame({
    "Month" : ["Jan", "Jan", "Jan", "Feb", "Feb"],
    "Product" : ["A", "A", "B", "A", "B"], 
    "Sales" : [120, 130, 90, 150, 110]
})

data 
```

```python
data_table = data.pivot_table(
    index="Month",
    columns="Product",
    values="Sales",
    aggfunc=["mean", "sum", "count", "max", "min"]
)

data_table
```
