## **`GroupBy & Aggregation in Pandas`**

**Grouping means splitting data into groups based on one or more columns.**

Example groups:
   - Department-wise
   - City-wise
   - Category-wise
   
> **We use the `groupby` method to split data into multiple groups so that we can apply some aggregation functions on it.**

**Aggregation means summarizing each group using functions.**

Common summaries:
   - Mean
   - Sum
   - Count
   - Max / Min

```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'Department' : ["IT", "IT", "HR", "HR", "Sales"],
    "Employee" : ["A", "B", "C", "D", "E"], 
    "Salary" : [60_000, 65_000, 50_000, 52_000, 70_000]
})
df
```

```python
df.groupby("Department")["Salary"].mean()
df.groupby("Department")["Salary"].min()
df.groupby("Department")["Salary"].max()
```

```python
df.groupby("Department")["Salary"].agg(["mean", "min", "max"])
df.groupby("Department")["Salary"].aggregate(["mean", "min", "max"])
```

```python
df.groupby("Department")["Salary"].agg(avg_salay="mean", min_salary="min", max_salary="max")
```

```python
df.groupby(['Department', 'Employee'])['Salary'].sum()
```

```python
df.groupby("Department").agg({
    "Salary" : ["mean", "max"], 
    "Employee" : "count"
})
```

```python
df.groupby('Department').agg(
    avg_salary=('Salary', 'mean'),
    max_salary=('Salary', 'max'),
    emp_count=('Employee', 'count')
)
```
