# Data Visualization - Seaborn

**Seaborn** is a **high-level Python data visualization library** built on top of **Matplotlib**.

It simplifies the process of creating **beautiful, informative, and statistically meaningful plots**.

| Library | Role |
|--------|------|
| **Matplotlib** | Low-level control for creating plots |
| **Seaborn** | High-level statistical visualization |

In simple terms:

Matplotlib → Full control over plots  
Seaborn → Faster and more beautiful statistical visualizations


## **Why Use Seaborn?**

Seaborn makes plots:

- **More beautiful**
- **More informative**
- **Easier to create**

It also provides **built-in statistical visualizations**, which makes it very powerful for **data analysis and EDA**.


```python
import seaborn as sns
import matplotlib.pyplot as plt
```

### **Seaborn provides built-in datasets for learning.**

```python
datasets = sns.get_dataset_names()
print(datasets)

tips = sns.load_dataset("tips")
print(type(tips))
tips.head()
```

### Data Layer
| Parameter                  | Used For                     | Example                        | Explanation                       |
| -------------------------- | ---------------------------- | ------------------------------ | --------------------------------- |
| **data**                   | Specifies the dataset        | `data=df`                      | The dataframe containing the data |
| **x**                      | Variable for x-axis          | `x="age"`                      | Column shown on horizontal axis   |
| **y**                      | Variable for y-axis          | `y="salary"`                   | Column shown on vertical axis     |

```python
sns.scatterplot(data=tips, x="total_bill", y="tip")
```

### Information Layer
| Parameter                  | Used For                     | Example                        | Explanation                       |
| -------------------------- | ---------------------------- | ------------------------------ | --------------------------------- |
| **hue**                    | Color grouping               | `hue="gender"`                 | Colors points based on category   |
| **style**                  | Different marker styles      | `style="smoker"`               | Changes shape of points           |
| **size**                   | Point size based on variable | `size="population"`            | Larger values → bigger points    

```python
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",
    size="size"
)
```

### Presentation Layer

| Parameter                  | Used For                     | Example                        | Explanation                       |
| -------------------------- | ---------------------------- | ------------------------------ | --------------------------------- |
| **palette**                | Color theme                  | `palette="Set2"`               | Controls colors of categories     |
| **figsize** *(Matplotlib)* | Figure size                  | `plt.figure(figsize=(8,5))`    | Controls plot size                |
| **title** *(Matplotlib)*   | Plot title                   | `plt.title("Sales vs Profit")` | Adds title to plot                |
| **xlabel** *(Matplotlib)*  | X-axis label                 | `plt.xlabel("Age")`            | Label for x-axis                  |
| **ylabel** *(Matplotlib)*  | Y-axis label                 | `plt.ylabel("Salary")`         | Label for y-axis                  |


```python
sns.set_theme() 

df = sns.load_dataset("tips")
plt.figure(figsize=(8, 5))

sns.relplot(
    data=df,
    x="total_bill",
    y="tip",
    kind="scatter",
    hue="smoker",
    size="size",
    style="smoker",
    palette="Set1"
)

plt.title("Tips Scatter Plot")
plt.xlabel("Total bill")
plt.ylabel("Tip")
plt.show()
```
