## **`Merging, Joining & Concatenation in Pandas`**

Imagine we work in a company.

we have:
- File 1 → Customer information
- File 2 → Orders information
- File 3 → Payments information

They are stored separately.

But for analysis, we must combine them.

That combining process = Merge / Join / Concatenation

---

```python
import pandas as pd
```

### **1️⃣ MERGE**

**Merge is used for SQL-like joins and combines two DataFrames based on common columns (keys).**

It supports: 

- Inner join - Only keep common values of both df
- Outer join - Keep all values of both df and replace missing with NaN
- Left join - Keep all values of left df and replace missing with NaN
- Right join - Keep all values of right df and replace missing with NaN

```python
customer_table = pd.DataFrame({
    "customer_id" : [1, 2], 
    "Name" : ["Ali", "Sara"]
})

orders_table = pd.DataFrame({
    "customer_id" : [1, 2, 3],
    "Order" : ["Laptop", "Phone", "PC"]
})
```

```python
customer_table
```

```python
orders_table
```

```python
pd.merge(customer_table, orders_table, on="customer_id")
```

```python
pd.merge(customer_table, orders_table, on="customer_id", how="inner")
```

```python
pd.merge(customer_table, orders_table, on="customer_id", how="outer")
```

```python
pd.merge(customer_table, orders_table, on="customer_id", how="left")
```

```python
pd.merge(customer_table, orders_table, on="customer_id", how="right")
```

---

### **2️⃣ JOIN (Index-Based Merge)**

**It is used to combine two DataFrames based on index by default & can also join on a key column.**

**SHORTCUT MERGE**

When to Use Join? 

- When both DataFrames already share the same index.
- When index is meaningful (like date index)

Important : `merge() is more flexible`  &  `join() is simpler but less powerful`


```python
dff1 = pd.DataFrame({
    "Name" : ["Ali", "Sara", "Zoya"],
}, index=[1, 2, 3])

dff2 = pd.DataFrame({
    "Marks" : [85, 90, 78],
}, index=[1, 2, 3])
```

```python
dff1
```

```python
dff2
```

```python
dff1.join(dff2)
```

---

### **3️⃣ Concatenation (Stacking Data)**

**Stacking DataFrames on top of each other (rows) or side by side (columns)**


```python
df1 = pd.DataFrame({
    "id" : [1, 2, 3], 
    "name" : ["Ali", "Sara", "Zoya"]
})

df2 = pd.DataFrame({
    "id" : [4, 5, 6], 
    "name" : ["Alia", "Hamza", "Nouman"]
})
```

```python
df1
```

```python
df2
```

- **Vertical Concatenation (default)**

Used when:

   Same columns

   More rows


```python
pd.concat([df1, df2]) # labels are preserved
```

```python
pd.concat([df1, df2], ignore_index=True)
```

- **Horizontal Concatenation**

Used when:

Same number of rows

Different columns

```python
pd.concat([df1, df2], axis=1)
```
