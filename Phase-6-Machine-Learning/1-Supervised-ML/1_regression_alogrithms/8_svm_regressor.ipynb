# SVM Regressor

```python
import pandas as pd 
from sklearn import datasets
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVR 
from sklearn.metrics import r2_score
from sklearn.pipeline import Pipeline
```

```python
df = datasets.load_diabetes(as_frame=True).frame

df.head()
```

```python
X = df.drop("target", axis=1) 
y = df["target"] 

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Kernel = Linear

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVR(kernel="linear", C=1))
])

pipeline.fit(X_train, y_train)

y_train_pred = pipeline.predict(X_train)
y_test_pred = pipeline.predict(X_test)

print("Training r2 score: ", r2_score(y_train, y_train_pred))
print("Testing r2 score: ", r2_score(y_test, y_test_pred))
```

```python
# Kernel = poly

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVR(kernel="poly", C=1))
])

pipeline.fit(X_train, y_train)

y_train_pred = pipeline.predict(X_train)
y_test_pred = pipeline.predict(X_test)

print("Training r2 score: ", r2_score(y_train, y_train_pred))
print("Testing r2 score: ", r2_score(y_test, y_test_pred))
```

```python
# Kernel = sigmoid

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVR(kernel="sigmoid", C=1))
])

pipeline.fit(X_train, y_train)

y_train_pred = pipeline.predict(X_train)
y_test_pred = pipeline.predict(X_test)

print("Training r2 score: ", r2_score(y_train, y_train_pred))
print("Testing r2 score: ", r2_score(y_test, y_test_pred))
```

```python
# Kernel = rbf

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVR(kernel="rbf", C=1))
])

pipeline.fit(X_train, y_train)

y_train_pred = pipeline.predict(X_train)
y_test_pred = pipeline.predict(X_test)

print("Training r2 score: ", r2_score(y_train, y_train_pred))
print("Testing r2 score: ", r2_score(y_test, y_test_pred))
```

## GridSearchCV for hyperparameter

```python
pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVR())
])

param_grid = {
    "svm__C": [1, 2, 5, 50, 100],
    "svm__kernel": ["linear", "rbf"],
    "svm__epsilon": [0.01, 0.1, 0.2, 0.3, 0.5]
}

grid_search = GridSearchCV(
    pipeline,
    param_grid,
    scoring="r2", 
    cv=5
)

grid_search.fit(X_train, y_train)

y_train_pred = grid_search.predict(X_train)
y_test_pred = grid_search.predict(X_test)

print("Training r2 score: ", r2_score(y_train, y_train_pred))
print("Testing r2 score: ", r2_score(y_test, y_test_pred))
```

```python
print("best params -- ", grid_search.best_params_)
```
