## `Cleaning Data` 

- ## **`Handling Missing Values 🐼`**

**In Pandas, missing data is usually represented as:**

- NaN (Not a Number)
- None
- NaT (for dates)

**Real-life reasons for missing data:**

- Data not collected
- Sensor failure
- User skipped a field
- Merging datasets

**Missing data is NORMAL, not an error.**

```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv") 
df
```

#### 1️⃣ **`isnull() / isna()`**

- Detecting missing values (ALWAYS first step)
- Returns True / False for every cell.

```python
df.isnull()
```

```python
df.isna()
```

#### 2️⃣ **`isnull().sum()`**

- Count missing values

```python
df.isnull().sum()
```

#### 3️⃣ **`dropna()`**

- Removing missing values (Drop)

> ***Drop rows with ANY missing value***

```python
df.dropna()
```

> ***Drop rows where ALL values are missing***

```python
df.dropna(how="all")
```

> ***Drop rows based on specific columns***

```python
df.dropna(subset=["age"])
```

#### 4️⃣ **`fillna(value)`**

- Filling missing values

> ***Fill with a constant value***

```python
df.fillna(0)
df.fillna("Unknown")
```

> ***Fill numeric columns with mean/ median***

```python
cleaned_data = df.copy()
cleaned_data["age"] = cleaned_data["age"].fillna(df["age"].mean())

cleaned_data
```

```python
df["age"].fillna(df["age"].median)
```

> ***Fill categorical data with mode***

```python
df["country"].fillna(df["country"].mode()[0])
```

#### 5️⃣ **`ffill()`**

- Forward fill (time-based data)

```python
df.ffill()
```

#### 6️⃣ **`bfill()`**

- backward fill (time-based data)

```python
df.bfill()
```

#### 7️⃣ **`Filling multiple columns differently`**

```python
df.fillna({
    "country" : df["country"].ffill(),
    "age" : df["age"].mean(),
    "gender" : "NA"
})
```

#### 8️⃣ **`In-place vs not`**

**What inplace=True actually does**

- Many Pandas methods (like fillna, drop, sort_values) can either:

   - Return a new DataFrame with changes applied.

   - Modify the original DataFrame directly.

`inplace=True` means:

"Do the operation **directly on the existing DataFrame**, instead of creating a copy."

- Avoid `inplace=True` in ML pipelines — assign back instead.

```python
# Fill numeric columns with a number
# df['age'].fillna(0, inplace=True)
# df['income'].fillna(0, inplace=True)

# # Fill object/string columns with "None"
# df['name'].fillna("None", inplace=True)
# df['country'].fillna("None", inplace=True)
# df['gender'].fillna("None", inplace=True)
```

- ## **`Handling Duplicate Values 🐼`**


- **Duplicate data = repeated rows or repeated values that should logically appear only once**

- Duplicates silently break analysis & ML models

| Task             | Code                                 |
| ---------------- | ------------------------------------ |
| Check duplicates | `df.duplicated()`                    |
| Count duplicates | `df.duplicated().sum()`              |
| View duplicates  | `df[df.duplicated()]`                |
| Drop duplicates  | `df.drop_duplicates()`               |
| Based on column  | `df.drop_duplicates(subset=["col"])` |
| Keep last        | `keep="last"`                        |
| Drop all         | `keep=False`                         |


#### 1️⃣ **`duplicated()`**


```python
# check duplicates
df.duplicated()
```

```python
# count duplicates
df.duplicated().sum()
```

```python
# view duplicates
df[df.duplicated()]
```

```python
# column wise
df[["country", "gender"]].duplicated()
```

#### 2️⃣ **`drop_duplicates`**


```python
# drop duplicates
df2 = df.copy()

df2.drop_duplicates(inplace=True)
df2
```

```python
# based on column
df.drop_duplicates(subset=["name"])
```

- ## **`Handling Data Types & data-time 🐼`**

- If data types are wrong:

  - Math fails

  - Sorting breaks

  - ML models reject data

- 80% bugs in ML pipelines come from wrong data types


####  **Common Wrong Data Type Problems**  |  **Golden Data Type Cleaning Strategy**


| Problem | Example | Correct Data Type | Best Method |
|----------|----------|------------------|-------------|
| Numbers stored as text | `"45"`, `"1000"` | Numeric | `pd.to_numeric()` |
| Dates stored as strings | `"2024-01-15"` | `datetime` | `pd.to_datetime()` |
| Mixed types in column | `10`, `"unknown"` | Numeric (cleaned) | `pd.to_numeric(errors="coerce")` |
| Boolean stored as text | `"yes"`, `"no"` | Boolean | `.map({"yes": True, "no": False})` |
| Repeated text categories | `"Male"`, `"Female"` | `category` | `.astype("category")` |

- ## **`Handling Strings 🐼`**


- **In Pandas:**

  **String operations are vectorized using .str**




```python
df
```

#### **1. Case Conversion**
| Method              | What it does         | Example                     |
| ------------------- | -------------------- | --------------------------- |
| `.str.lower()`      | Convert to lowercase | `"Ali"` → `"ali"`           |
| `.str.upper()`      | Convert to uppercase | `"Ali"` → `"ALI"`           |
| `.str.title()`      | Title case           | `"john doe"` → `"John Doe"` |
| `.str.capitalize()` | First letter capital | `"ali"` → `"Ali"`           |
| `.str.swapcase()`   | Swap upper/lower     | `"Ali"` → `"aLI"`           |


```python
df["gender"].str.capitalize()
```

#### **2. Removing & Trimming Spaces**
| Method          | What it does               | Example             |
| --------------- | -------------------------- | ------------------- |
| `.str.strip()`  | Remove spaces (both sides) | `" Ali "` → `"Ali"` |
| `.str.lstrip()` | Remove left spaces         | `" Ali"` → `"Ali"`  |
| `.str.rstrip()` | Remove right spaces        | `"Ali "` → `"Ali"`  |



```python
df["name"].str.strip()
```

#### **3. Replace & Clean Text**
| Method                | What it does  | Example               |
| --------------------- | ------------- | --------------------- |
| `.str.replace(a, b)`  | Replace text  | `"Karachi"` → `"KHI"` |
| `.str.replace(regex)` | Regex replace | Remove extra spaces   |


```python
df["country"].str.replace("USA", "Pakistan")
```

#### **4. Searching & Matching (Filtering Helper)**
| Method              | Use case         | Returns |
| ------------------- | ---------------- | ------- |
| `.str.contains()`   | Check substring  | Boolean |
| `.str.startswith()` | Starts with text | Boolean |
| `.str.endswith()`   | Ends with text   | Boolean |
| `.str.match()`      | Full regex match | Boolean |
| `.str.findall()`    | Find all matches | List    |


```python
df["country"].str.contains("india", case=False)
```

#### **5. Splitting & Combining Strings**

| `.str.join()`      | Join characters     | `"abc"` → `"a-b-c"`   |
| `.str.cat()`       | Concatenate columns | Combine names         |
