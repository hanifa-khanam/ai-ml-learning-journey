## **Line Plot**

A line plot connects data points using straight lines.

**Used for:**

 - Time series

 - Trends

 - Continuous numerical change

 - Model training curves

###  Key Features of a Line Plot

| Feature | Description | Example |
|----------|-------------|----------|
| **X-axis** | Usually continuous values | Time, sequence, measurements |
| **Y-axis** | Values being measured | Sales, temperature, marks |
| **Line Style** | Controls the appearance of the line | `-` (solid), `--` (dashed), `:` (dotted) |
| **Markers** | Shows individual data points | `o`, `s`, `^`, `*` |

```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25] 

plt.plot(x, y)
plt.xlabel("X values")
plt.ylabel("y values")
plt.title("Simple line Plot")
plt.grid(True)  # adds a grid
plt.show()
```

- ### **Styling a Line Plot**

**Parameter:**

`color`    (line color)

`linestyle`   (solid/dashed) ` '-', '--', '-.', ':', 'None', ' ', '', 'solid', 'dashed', 'dashdot', 'dotted'`

`linewidthmarker`   (Thickness)

`markersize`

`label`   (for legend)


```python
plt.plot(x, y, color="blue", linestyle=":", linewidth=2, marker="<", markersize=6 ) 
plt.show()
```

- ### **Format Strings (Shortcut Styling)**

A format string consists of a part for color, marker and line: 

`fmt = [color] [marker] [linestyle]`

```python
plt.plot(x,y, 'go--')
```

- ### **Multiple Datasets on Line Plot**

```python
y2 = [1, 3, 5, 7, 9] 

plt.plot(x, y, label="Product A") 
plt.plot(x, y2, label="Product B") 

plt.legend()
```

### Useful `matplotlib.pyplot` Styling & Layout Functions

| Function | Use | 
|----------|------|
| `plt.style.available` | Shows all available plotting styles | 
| `plt.style.use("style_name")` | Applies a specific style to the plot | 
| `plt.grid()` | Adds grid lines to the plot |
| `plt.tight_layout()` | Adjusts spacing automatically |
| `plt.savefig("abc.png")` | Saves the figure to a file |
